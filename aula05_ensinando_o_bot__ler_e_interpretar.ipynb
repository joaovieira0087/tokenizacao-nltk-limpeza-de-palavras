```python
from tkinter.constants import WORD
import nltk
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords

nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')

frase = ("O silêncio às vezes diz mais do que mil palavras mal formuladas.")

#tokenização - cortar a frase em pedaços(token)

tokens = word_tokenize(frase.lower())

# criando uma lista só com letra limpara a pontuação da nossa frase
token_limpos = [t for t in tokens if t.isalpha()]

palavras_inuteis = set(stopwords.words('portuguese'))
frase_limpa = [palavra for palavra in token_limpos if palavra not in palavras_inuteis]

print(frase_limpa)
```

```python
import spacy

# !python -m spacy download pt_core_news_sm

nlp = spacy.load("pt_core_news_sm")

frase = "O suporte do SENAI SÃO PAULO e excelente! O atendimento da Marta foi rápido hoje."

frase_processada = nlp(frase)

for ent in frase_processada.ents:
  print(f"Entidades encontrada:{ent.text} ==> Tipo:{ent.label_}")
```
