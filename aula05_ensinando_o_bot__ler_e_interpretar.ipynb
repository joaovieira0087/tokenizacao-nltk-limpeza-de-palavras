```python
from tkinter.constants import WORD
import nltk
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords

nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')

frase = ("llll")

#tokenização - cortar a frase em pedaços(token)

tokens = word_tokenize(frase.lower())

# criando uma lista só com letra limpara a pontuação da nossa frase
token_limpos = [t for t in tokens if t.isalpha()]

palavras_inuteis = set(stopwords.words('portuguese'))
frase_limpa = [palavra for palavra in token_limpos if palavra not in palavras_inuteis]

print(frase_limpa)
```
